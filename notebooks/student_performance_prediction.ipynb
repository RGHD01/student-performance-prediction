<h3>Identify Outliers <h3>

```python
#absences
df['absences'].value_counts()
```

```python
plt.figure(figsize=(7,7))
sns.boxplot(df['absences'].values,orient='h',)
```

<h3>Remove Outliers</h3>

```python
# TODO:  Remove Outliers
Q1 = np.percentile(df['absences'], 25)
Q3 = np.percentile(df['absences'], 75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df['absences'] = df['absences'].clip(lower=0, upper=upper_bound)



plt.figure(figsize=(5,5))
sns.boxplot(df['absences'].values, orient='h')
```

#Choose and  fit model




```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
data= pd.read_csv('/content/cleaned_student_performance2 (3).csv')
```

```python
data.columns
```

```python
features= ['failures','absences','studytime','G1','G2']
target=['G3']
```

```python
x=data[features]
y=data[target]
```

```python
x_train, x_test, y_train,y_test= train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
y_predict=model.predict(x_test)
```

```python
x_test[1:2]
```

```python
y_predict[1:2]
```

```python
y_test[1:2]
```

```python
m_error= mean_squared_error(y_test, y_predict)
r2= r2_score(y_test,y_predict)

print(f"Mean Squared Error: {m_error:.2f}")
print(f"R-squared: {r2:.2f}")
```

```python
accuracy =r2_score(y_test,y_predict) *100
print(f"Model Accuracy: {accuracy:.2f}%")
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_predict, alpha=0.7, color='b')
plt.xlabel('Actual G3')
plt.ylabel('Predicted Prices')
plt.title('Actual vs Predicted G3')
plt.show()
```
